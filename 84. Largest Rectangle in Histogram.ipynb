```python
heights = [2,1,5,6,2,3]
heights = [2,4]
heights = [3,6,5,7,4,8,1,0]
# heights = [0,1,0,2,1,0,1,3,2,1,2,1]
```

```python
def sol(heights):

    heights.append(-1)
    print(heights)

    n = len(heights)

    stack = []
    limits = {}
    best = -1

    for i in range(n):

        print(i)
        print('stack in', stack)

        while stack and heights[i] < heights[stack[-1]]:

            lastIdx = stack[-1]
            lastVal = heights[stack[-1]]

            newIdx = i
            newVal = heights[i]
            
            # сохраняю в словарь наибольший лимит для числа
            if not lastVal in limits:
                limits[lastVal] = [0,0]
            limits[lastVal][1] = max(limits[lastVal][1], i - lastIdx)

            # меньшее забирает все что было у последнего
            if not newVal in limits:
                limits[newVal] = [0,0]
            limits[newVal][0] = max(limits[newVal][0], limits[lastVal][0] + limits[lastVal][1])

            stack.pop()

        if not stack or heights[i] >= heights[stack[-1]]:
            stack.append(i)

        print('stack out', stack)


        # если пришло меньшее число, то все что было до больше - все это тоже доступная ширина


        res = -1
        for num, limit in limits.items():
            area = num * (limit[0] + limit[1])
            res = max(res, area)
            print('-', num, limit, area)
        
    return res
              
```

```python
def sol(heights):

    heights.append(0)
    stack = []
    best = -1

    
    for i in range(len(heights)):
    
        while stack and heights[i] <= heights[stack[-1]]:  

            height = heights[stack.pop()]
            width = (i - 1) - stack[-1] if stack else i
        
            area = height * width
            best = max(best, area)

        
        stack.append(i)


    return best
```
