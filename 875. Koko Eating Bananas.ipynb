```python
import math 
def sol(piles, h):

    minK, maxK = 1, max(piles)

    while minK <= maxK:

        k = (minK + maxK) // 2
        print('k', k)

        hoursNeed = 0
        for pile in piles:
            hoursNeed += math.ceil(pile/k)
            print(pile, math.ceil(pile/k), 'for pile in piles')

        print('hoursNeed', hoursNeed)

        if hoursNeed < h: # если при таком k понадобилось времени меньше чем h - то можно есть медленее - k нужно искать в меньшей  половине
            maxK = k - 1
        elif hoursNeed > h: # то есть не успеваем съесть - k нужно искать в большей половине
            minK = k + 1
        else:
            return k
```

```python
def sol(piles, h):

    l, r = 1, max(piles)
    res = r

    while l <= r:

        k = (l + r) // 2

        hours = 0
        for pile in piles:
            hours += math.ceil(pile/k)

        if hours <= h:
            res = min(res, k) # вот тут не догадался
            r = k - 1
        else: 
            l = k + 1


    return res
```
