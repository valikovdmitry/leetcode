```python
nums = [4,5,6,7,0,1,2]
nums = [3,4,5,1,2]
nums = [11,13,15,17]
nums = [2,1]
```

```python
def sol(nums):

    l, r = 0, len(nums) - 1
    res = nums[l]

    while l <= r:
        print(l,r)
        
        # значит пивота нет, значит сортированная - 
        # отдаю 0 элемент
        if nums[l] < nums[r]: 
            res = min(res, nums[l])
            break
           
        m = (l+r) // 2

        # значит пивота еще не было - ищем справа
        if nums[m] >= nums[0]: 
            l = m + 1
        
        # значит пивот уже случился - ищем слева
        elif nums[m] < nums[0]: 
            res = min(res, nums[m])
            r = m - 1


    return res
```

```python
sol(nums)
```
