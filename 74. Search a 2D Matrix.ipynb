```python
matrix = [[1,3,5,7],[10,11,16,20],[23,30,34,60]]
target = 13
```

```python
def sol(matrix, target):
    
    nums = []
    for row in matrix:
        for num in row:
            nums.append(num)

    l,r = 0, len(nums)-1

    while l <= r:
        mid = (l+r)//2

        if nums[mid] > target:
            r = mid - 1
        elif nums[mid] < target:
            l = mid + 1
        else:
            return True
            

    return False
```

```python
sol(matrix, target)
```

```python
matrix = [[1, 3, 5, 7 ],
          [10,11,16,20],
          [23,30,34,60]]
target = 4
```

```python
def sol(matrix, target):

    top, down = 0, len(matrix)-1

    while top <= down:

        mid = (top + down) // 2
        row = matrix[mid]

        if row[0] > target:
            down = mid - 1
        elif row[-1] < target:
            top = mid + 1
        else:  
            
            l,r  = 0, len(row) - 1
            while l <= r:
                m = (l+r)//2

                if row[m] > target:
                    r = m-1
                elif row[m] < target:
                    l = m + 1
                else:
                    return True

            return False
    
    return False # отработать вариант где всего одна строка
        
```

```python
sol(matrix, target)
```

```python
matrix = [[1, 3, 5, 7 ],
          [10,11,16,20],
          [23,30,34,60]]
target = 3
```

```python
def sol(matrix, target):

    ROWS, COLS = len(matrix), len(matrix[0])

    top, bot = 0, ROWS - 1
    while top <= bot:
    
        row = (top + bot) // 2
    
        if matrix[row][0] > target:
            bot = row - 1
        elif matrix[row][-1] < target:
            top = row + 1
        else:  
            break 
    
    if not (top <= bot):
        return False
    
    row = (top + bot) // 2
    
    l,r  = 0, COLS - 1
    while l <= r:
        m = (l+r)//2
    
        if matrix[row][m] > target:
            r = m - 1
        elif matrix[row][m] < target:
            l = m + 1
        else:
            return True
    
    return False
```

```python
sol(matrix, target)
```
