```python
def search(nums, target):

    l, r = 0, len(nums) - 1

    while l < r:
        mid = (l + r) // 2

        if nums[mid] > target:
            r = mid

        elif nums[mid] < target:
            l = mid

        else:
            return mid

    return -1
```

```python
search(nums, target)
```

```python
nums = [2]
target = 2
```

```python
def search(nums, target):

    l, r = 0, len(nums) - 1

    while l <= r:

        mid = (l+r)//2

        if nums[mid] > target:
            r = mid - 1
        elif nums[mid] < target:
            l = mid + 1
        else:
            return mid

    
    return -1
```

```python
search(nums, target)
```
