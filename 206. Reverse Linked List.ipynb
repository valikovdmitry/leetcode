```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
    
    def __repr__(self):
        return str(self.__dict__)
        
class Solution:
    def reverseList(self, head):
        
        prev = None
        curr = head
        
        while curr:

            next_node = curr.next  
            curr.next = prev
            prev = curr
            curr = next_node
            

        return prev

```
